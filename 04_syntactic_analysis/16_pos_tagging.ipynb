# 16. Part-of-Speech Tagging

## Overview
Part-of-speech (POS) tagging assigns a grammatical category to each token, such as noun, verb, adjective, or adverb. POS tags provide useful signals for parsing, information extraction, and text analysis.

This notebook introduces POS tagging with NLTK, explains common tag conventions, and demonstrates how tags can be inspected and summarized.

## 1. Setup

The examples use NLTK's English tokenizer and pretrained POS tagger. Run the downloads below once in an environment that has internet access.

```python
import nltk

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
```

## 2. Tokenize and tag a sentence

`nltk.pos_tag` returns `(token, tag)` pairs. The default English tagset is Penn Treebank.

```python
from nltk import word_tokenize, pos_tag

sentence = "The curious researcher carefully analyzed the financial reports."
tokens = word_tokenize(sentence)
tagged_tokens = pos_tag(tokens)

print(tagged_tokens)
```

## 3. Read common Penn Treebank tags

A few common tags are shown below. Tag meanings depend on context, so a tag should be interpreted together with its token and sentence.

```python
tag_examples = {
    "NN": "noun, singular or mass",
    "NNS": "noun, plural",
    "VB": "verb, base form",
    "VBD": "verb, past tense",
    "JJ": "adjective",
    "RB": "adverb",
    "DT": "determiner",
    "IN": "preposition or subordinating conjunction",
    "PRP": "personal pronoun",
    "CC": "coordinating conjunction",
}

for tag, description in tag_examples.items():
    print(f"{tag:>4}  {description}")
```

## 4. Inspect tokens by grammatical category

Filtering by tag prefix is a convenient first analysis. For example, Penn Treebank noun tags usually begin with `NN`, while adjective tags begin with `JJ`.

```python
nouns = [word for word, tag in tagged_tokens if tag.startswith("NN")]
verbs = [word for word, tag in tagged_tokens if tag.startswith("VB")]
adjectives = [word for word, tag in tagged_tokens if tag.startswith("JJ")]
adverbs = [word for word, tag in tagged_tokens if tag.startswith("RB")]

print("Nouns:", nouns)
print("Verbs:", verbs)
print("Adjectives:", adjectives)
print("Adverbs:", adverbs)
```

## 5. Convert to the universal tagset

The universal tagset groups detailed tags into a smaller set such as `NOUN`, `VERB`, `ADJ`, and `ADV`. This can make summaries easier to compare.

```python
universal_tags = pos_tag(tokens, tagset="universal")
print(universal_tags)
```

## 6. Count tag frequencies

Tag frequencies provide a quick overview of the grammatical structure in a passage.

```python
from collections import Counter

tag_counts = Counter(tag for _, tag in tagged_tokens)
for tag, count in tag_counts.most_common():
    print(f"{tag:>5}: {count}")
```

## 7. Tag a longer text

This example processes a short paragraph as a sequence of tokens. POS tagging is context-sensitive: the same word can receive different tags in different sentences.

```python
text = (
    "Markets reacted quickly after the central bank announced its decision. "
    "Investors watched prices closely and analysts discussed the possible effects."
)

text_tokens = word_tokenize(text)
text_tagged = pos_tag(text_tokens)

for token, tag in text_tagged:
    print(f"{token:<14} {tag}")
```

## Summary

- POS tagging assigns grammatical labels to tokens.
- NLTK's `pos_tag` uses the Penn Treebank tagset by default.
- The universal tagset offers a smaller, more general inventory.
- A tag is a model prediction, not a guarantee; ambiguity and domain-specific language can affect results.
