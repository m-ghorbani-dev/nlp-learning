# Sentence Tokenization

## Introduction

Sentence Tokenization is the process of splitting a text into individual sentences.

It is an important step in Natural Language Processing (NLP).

## Goals

In this notebook, we will:
- Split text into sentences.
- Count the number of sentences.
- Display each sentence with a number.
- Count the words in each sentence.
- Understand why splitting on periods alone is not always reliable.

## Library

We will use NLTK for sentence tokenization.

```python
# If NLTK is not installed, run this in a separate cell:
# %pip install nltk

import nltk
from nltk.tokenize import sent_tokenize

# Download the resources required for sentence tokenization
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
```

```python
text = """
Natural Language Processing (NLP) helps computers work with human language.
It is used in chatbots, search engines, and sentiment analysis.
For example, a model can classify a review as positive or negative!
How can we prepare text for these tasks?
Sentence tokenization is one useful first step.
"""

print(text)
```

```python
sentences = sent_tokenize(text)

print(sentences)
```

```python
print(type(sentences))
```

```python
for number, sentence in enumerate(sentences, start=1):
    print(f"{number}. {sentence}")
```

```python
sentence_count = len(sentences)

print("Number of sentences:", sentence_count)
```

```python
for number, sentence in enumerate(sentences, start=1):
    words = sentence.split()
    print(f"Sentence {number}: {len(words)} words")
```

```python
import nltk
from nltk.tokenize import sent_tokenize

# Sample text
text = """
Python is a popular programming language.
It is widely used in Natural Language Processing (NLP).
NLP helps computers analyze human language!
Would you like to learn more about NLP?
"""

# Split text into sentences
sentences = sent_tokenize(text)

# Display results
print("SENTENCE ANALYSIS")
print("-" * 30)

for number, sentence in enumerate(sentences, start=1):
    word_count = len(sentence.split())

    print(f"Sentence {number}: {sentence}")
    print(f"Word count: {word_count}")
    print()

print("Total sentences:", len(sentences))
```
