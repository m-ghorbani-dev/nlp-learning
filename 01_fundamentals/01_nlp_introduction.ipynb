# Introduction to Natural Language Processing (NLP)

## What is NLP?

Natural Language Processing (NLP) is a branch of Artificial Intelligence
that enables computers to process, analyze, and understand human language.

NLP combines concepts from:
- Artificial Intelligence
- Machine Learning
- Linguistics
- Deep Learning

## Applications of NLP

- Sentiment Analysis
- Text Classification
- Machine Translation
- Chatbots
- Named Entity Recognition (NER)
- Spam Detection
- Text Summarization
- Question Answering

```python
text = """
Natural Language Processing is an important field of Artificial Intelligence.
NLP helps computers process and analyze human language.
Python is widely used for NLP applications.
"""

print(text)
```

```python
print("Data type:", type(text))
print("Number of characters:", len(text))
```

```python
lower_text = text.lower()

print(lower_text)
```

```python
upper_text = text.upper()

print(upper_text)
```

```python
sample_text = "   Natural Language Processing is interesting.   "

clean_text = sample_text.strip()

print("Original:", repr(sample_text))
print("Cleaned:", repr(clean_text))
```

```python
sentence = "Natural Language Processing is interesting"

words = sentence.split()

print(words)
print("Number of words:", len(words))
```

```python
text = """
Python is useful for NLP.
Python is also useful for Machine Learning.
NLP helps computers understand text.
"""

words = text.lower().split()

print("Words:", words)
print("Total words:", len(words))
```

```python
from collections import Counter

text = """
Python is useful for NLP.
Python is useful for Machine Learning.
NLP is useful for text processing.
"""

words = text.lower().split()

word_counts = Counter(words)

print(word_counts)
```

```python
print(word_counts.most_common(5))
```

```python
text = "Hello!!! NLP is interesting... Isn't it?"

print("Original text:")
print(text)

clean_text = text.replace("!", "").replace(".", "")

print("\nText after removing ! and .:")
print(clean_text)
```

```python
def preprocess_text(text):
    text = text.lower()
    text = text.strip()
    text = text.split()

    return text


sample_text = "   Python is GREAT for NLP   "

processed_text = preprocess_text(sample_text)

print(processed_text)
```

```python
text = """
Python is useful.
Python is powerful.
NLP is useful.
"""

words = text.lower().split()

unique_words = set(words)

print("All words:", words)
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```
