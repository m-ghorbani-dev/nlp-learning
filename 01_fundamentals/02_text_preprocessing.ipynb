# Text Preprocessing in NLP

## Introduction

Text preprocessing is the process of cleaning and transforming raw text
into a suitable format for analysis and machine learning.

Raw text may contain:
- Unnecessary whitespace
- Inconsistent capitalization
- Punctuation
- Special characters
- HTML tags
- URLs
- Numbers
- Newline characters

## Goals

In this notebook, we will learn how to:

1. Convert text to lowercase
2. Remove leading and trailing whitespace
3. Normalize whitespace
4. Remove punctuation
5. Remove digits when appropriate
6. Handle newline and tab characters
7. Remove URLs
8. Remove HTML tags
9. Build a reusable preprocessing function
10. Compare raw and cleaned text

**Important:** Preprocessing depends on the NLP task.
We should not remove every punctuation mark, number, or special character
without considering the meaning of the text.

```python
import re
import string
from collections import Counter


def preprocess_text(text):

    # Remove 
    text = text.lower()
    text = re.sub(r"https?://\S+", "", text)
    text = re.sub(r"<[^>]+>", "", text)
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )
    

    # 5. Normalize whitespace
    text = " ".join(text.split())

    return text


def analyze_text(text):
    words = text.split()

    return {
        "characters": len(text),
        "words": len(words),
        "unique_words": len(set(words)),
        "most_common_words": Counter(words).most_common(5)
    }


# Raw text
raw_text = """
    <p>Python is AMAZING!!!</p>
    Python is useful for NLP.
    Learn NLP at https://example.com
    NLP is interesting in 2026.
"""

# Preprocessing
cleaned_text = preprocess_text(raw_text)

# Analysis
statistics = analyze_text(cleaned_text)

# Results
print("Original text:")
print(raw_text)

print("\nCleaned text:")
print(cleaned_text)

print("\nText statistics:")
for key, value in statistics.items():
    print(f"{key}: {value}")
```
