# Word Tokenization

## Introduction

Word Tokenization is the process of splitting text into individual words
and tokens.

A token can be a word, punctuation mark, number, or another text element,
depending on the tokenizer.

## Goals

In this notebook, we will learn how to:

- Split text into words using Python.
- Tokenize text using NLTK.
- Compare different tokenization methods.
- Handle punctuation.
- Count tokens and unique tokens.
- Understand why tokenization matters in NLP.

```python
# If NLTK is not installed, run this in a separate cell:
# %pip install nltk

import nltk
from nltk.tokenize import word_tokenize, TreebankWordTokenizer

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
```

```python
text = """
Natural Language Processing (NLP) is interesting!
Python is widely used for NLP.
Isn't tokenization important? Yes, it is.
"""

print(text)
```

```python
simple_tokens = text.split()

print(simple_tokens)
print("Number of tokens:", len(simple_tokens))
```

```python
tokens = word_tokenize(text)

print(tokens)
print("Number of tokens:", len(tokens))
```

```python
print("Using split():")
print(text.split())

print("\nUsing word_tokenize():")
print(word_tokenize(text))
```

```python
tokenizer = TreebankWordTokenizer()

treebank_tokens = tokenizer.tokenize(text)

print(treebank_tokens)
print("Number of tokens:", len(treebank_tokens))
```

```python
tokens = word_tokenize(text)

words = [
    token for token in tokens
    if any(char.isalnum() for char in token)
]

punctuation = [
    token for token in tokens
    if not any(char.isalnum() for char in token)
]

print("Words and alphanumeric tokens:", words)
print("Punctuation tokens:", punctuation)

print("Number of word-like tokens:", len(words))
print("Number of punctuation tokens:", len(punctuation))
```

```python
from collections import Counter

tokens = word_tokenize(text.lower())

words = [
    token for token in tokens
    if any(char.isalnum() for char in token)
]

word_counts = Counter(words)

print("Word frequencies:")
print(word_counts)

print("\nMost common words:")
print(word_counts.most_common(5))
```

```python
import nltk
from nltk.tokenize import word_tokenize
from collections import Counter

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)


def analyze_text(text):
    tokens = word_tokenize(text)
    words = [
        token.lower()
        for token in tokens
        if any(char.isalnum() for char in token)
    ]

    return {
        "tokens": tokens,
        "word_count": len(words),
        "unique_word_count": len(set(words)),
        "most_common_words": Counter(words).most_common(5)
    }


text = """
Python is useful for NLP.
NLP is interesting!
Python supports many NLP applications.
"""

result = analyze_text(text)

print("Tokens:")
print(result["tokens"])

print("\nWord count:")
print(result["word_count"])

print("\nUnique word count:")
print(result["unique_word_count"])

print("\nMost common words:")
print(result["most_common_words"])
```
