# Stop Words in NLP

## Project: Text Cleaning with Stop Words

Stop words are common words that often carry less useful information for certain NLP tasks.

Examples in English:
- the
- is
- a
- an
- and
- of

In this project, we will use NLTK to identify and remove English stop words.

We will also compare the original text with the cleaned text.

```python
text = """
Natural Language Processing is a field of Artificial Intelligence.
It helps computers understand and process human language.
The field is used in chatbots, search engines, and text classification.
"""

print(text)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")

stop_words = set(stopwords.words("english"))

print("Number of stop words:", len(stop_words))
print("Some stop words:", list(stop_words)[:20])
```

```python
tokens = word_tokenize(text)

filtered_words = [
    word for word in tokens
    if word.lower() not in stop_words
]

print("Original tokens:")
print(tokens)

print("\nFiltered tokens:")
print(filtered_words)
```

```python
original_count = len(tokens)
filtered_count = len(filtered_words)

print("Original word count:", original_count)
print("Word count after removing stop words:", filtered_count)
print("Removed words:", original_count - filtered_count)
```

```python
print("Original text:")
print(" ".join(tokens))

print("\nText after removing stop words:")
print(" ".join(filtered_words))
```

```python
def remove_stopwords(text, language="english"):
    stop_words = set(stopwords.words(language))
    tokens = word_tokenize(text)

    filtered_words = [
        word for word in tokens
        if word.lower() not in stop_words
    ]

    return filtered_words
```

```python
cleaned_tokens = remove_stopwords(text)

print(cleaned_tokens)
```

```python
def analyze_text(text):
    tokens = word_tokenize(text)
    cleaned_tokens = remove_stopwords(text)

    return {
        "original_word_count": len(tokens),
        "cleaned_word_count": len(cleaned_tokens),
        "removed_word_count": len(tokens) - len(cleaned_tokens),
        "cleaned_text": " ".join(cleaned_tokens)
    }


result = analyze_text(text)

for key, value in result.items():
    print(f"{key}: {value}")
```
