# Regular Expressions (Regex)

## Project: Spam Text Analyzer

In this project, we will analyze a text containing spam-related information.

We will learn how to:

- Search for patterns using re.search()
- Find all matches using re.findall()
- Use character classes
- Match words using special characters
- Extract email addresses
- Extract groups using group()
- Ignore uppercase and lowercase differences
- Use NLTK RegexpTokenizer

```python
text = """
Email spam, also known as junk email, is a type of electronic spam
where unsolicited messages are sent by email.

Spam messages are often commercial in nature.
Some messages contain disguised links that lead to phishing websites
or websites hosting malware.

For more information, contact:
barbara-alice95@gmail.com
Alice-barabara@yahoo.com
"""

print(text)
```

```python
import re

match = re.search(r"spam", text)

if match:
    print("Match found:", match.group())
else:
    print("No match found")
```

```python
match = re.search(r"food", text)

if match:
    print(match.group())
else:
    print("No match found")
```

```python
matches = re.findall(r"spam", text)

print(matches)
print("Number of matches:", len(matches))
```

```python
matches = re.findall(r"[sS]pam", text)

print(matches)
```

```python
matches = re.findall(r"spam", text, re.IGNORECASE)

print(matches)
```

```python
patterns = [
    r"[a-z]mail",
    r"[a-zA-Z]mail",
    r"\wmail",
    r".mail"
]

for pattern in patterns:
    print(f"Pattern: {pattern}")
    print(re.findall(pattern, text))
    print("-" * 40)
```

```python
emails = re.findall(r"[\w-]+@[\w.]+", text)

print(emails)
```

```python
match = re.search(r"([\w-]+)@([\w.]+)", text)

if match:
    print("Full email:", match.group(0))
    print("Username:", match.group(1))
    print("Domain:", match.group(2))
```

```python
from nltk.tokenize import RegexpTokenizer

tokenizer = RegexpTokenizer(r"[Ss]pam")

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
tokenizer = RegexpTokenizer(r"\w+")

tokens = tokenizer.tokenize(text)

print(tokens[:20])
```

```python
import re
from nltk.tokenize import RegexpTokenizer


def analyze_spam_text(text):
    # Find all occurrences of spam, ignoring case
    spam_words = re.findall(r"spam", text, re.IGNORECASE)

    # Extract email addresses using our educational pattern
    emails = re.findall(r"[\w-]+@[\w.]+", text)

    # Extract word-like tokens
    tokenizer = RegexpTokenizer(r"\w+")
    tokens = tokenizer.tokenize(text)

    return {
        "spam_count": len(spam_words),
        "emails": emails,
        "word_count": len(tokens),
        "unique_words": len(set(token.lower() for token in tokens))
    }


result = analyze_spam_text(text)

print("SPAM TEXT ANALYSIS")
print("-" * 30)

print("Spam occurrences:", result["spam_count"])
print("Email addresses:", result["emails"])
print("Word count:", result["word_count"])
print("Unique words:", result["unique_words"])
```
