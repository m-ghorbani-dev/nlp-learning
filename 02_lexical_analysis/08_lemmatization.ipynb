## 1. What Is Lemmatization?

Lemmatization attempts to find a word's dictionary base form, called its **lemma**.

Examples:
- `cooking` → `cooking` (as a noun, depending on context)
- `believes` → `believe`
- `better` → `good` when the correct POS and lexical rules are available

The result depends on the word and its grammatical role. Lemmatization is not simply the removal of a suffix.

### Stemming vs. Lemmatization
- **Stemming** applies rules that remove or modify word endings. Its output may not be a real word.
- **Lemmatization** uses lexical knowledge to return a dictionary form when possible.

## 2. Import the Required Tools

```python
# Import NLTK's WordNetLemmatizer and PorterStemmer
from nltk.stem import WordNetLemmatizer, PorterStemmer

# Create reusable lemmatizer and stemmer objects
lemmatizer = WordNetLemmatizer()
porter = PorterStemmer()
```

## 3. Basic Lemmatization

`WordNetLemmatizer` uses WordNet, a lexical database for English.

By default, `lemmatize()` treats a word as a noun (`pos="n"`).
This is important because the same word can have different base forms depending on its part of speech.

```python
# Define sample words
words = ["cooking", "believes", "studies", "cars", "children"]

# Lemmatize each word using the default noun POS
for word in words:
    lemma = lemmatizer.lemmatize(word)

    # Print the original word and its lemma
    print(f"{word:12} -> {lemma}")
```

## 4. The Importance of Part of Speech (POS)

The `pos` argument specifies the grammatical category:
- `"n"`: noun
- `"v"`: verb
- `"a"`: adjective
- `"r"`: adverb

If the POS is not specified, the lemmatizer assumes that the word is a noun. Therefore, verb forms such as `believes` may not be reduced as expected.

```python
# Compare lemmatization with different POS settings
pos_examples = [
    ("believes", "v"),
    ("studies", "v"),
    ("cars", "n"),
    ("better", "a"),
    ("quickly", "r"),
]

for word, pos in pos_examples:
    lemma = lemmatizer.lemmatize(word, pos=pos)

    # Show the word, POS tag, and resulting lemma
    print(f"{word:12} | POS: {pos} | Lemma: {lemma}")
```

## 5. Compare Default POS with Verb POS

This example demonstrates how the same word can produce different results depending on whether it is treated as a noun or a verb.

```python
# Words that can be interpreted as verbs
verb_words = ["cooking", "believes", "studies", "running", "played"]

for word in verb_words:
    # Default behavior treats the word as a noun
    noun_lemma = lemmatizer.lemmatize(word)

    # Explicitly treat the word as a verb
    verb_lemma = lemmatizer.lemmatize(word, pos="v")

    print(f"Word: {word:12} | Default: {noun_lemma:12} | Verb: {verb_lemma}")
```

## 6. Lemmatization vs. Stemming

Compare `WordNetLemmatizer` and `PorterStemmer` on the same list of words.

Keep in mind that these tools have different goals, so their outputs are not expected to match.

```python
# Define words for comparison
comparison_words = [
    "information",
    "cooking",
    "believes",
    "studies",
    "connected",
    "running",
    "better",
]

for word in comparison_words:
    # Use the default noun POS for a simple baseline
    lemma = lemmatizer.lemmatize(word)

    # PorterStemmer applies stemming rules
    stem = porter.stem(word)

    print(f"Word: {word:12} | Lemma: {lemma:12} | Stem: {stem}")
```

## 7. Lemmatizing a Sentence

A basic workflow is to tokenize a sentence and then lemmatize each token.
This example uses `word_tokenize`, which may require the appropriate NLTK tokenizer data in your environment.

```python
# Import the tokenizer
from nltk.tokenize import word_tokenize

# Define a sample sentence
sentence = "The children are playing and the dogs were running."

# Tokenize the sentence into words and punctuation
tokens = word_tokenize(sentence)

# Lemmatize each token using the default noun POS
lemmas = [lemmatizer.lemmatize(token) for token in tokens]

print("Tokens: ", tokens)
print("Lemmas: ", lemmas)
```
