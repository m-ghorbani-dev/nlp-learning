## 2. Find Synsets

A **synset** (synonym set) represents a specific meaning of a word. One word can belong to multiple synsets because it may have several meanings.

The `synsets()` function returns the synsets associated with a word. You can optionally provide a part-of-speech filter.

```python
# Find all WordNet synsets for the word "bank"
bank_synsets = wn.synsets("bank")

# Display the synsets
for synset in bank_synsets:
    print(synset)
```

### Filter synsets by part of speech

WordNet uses these common POS codes:
- `wn.NOUN` — noun
- `wn.VERB` — verb
- `wn.ADJ` — adjective
- `wn.ADV` — adverb

```python
# Retrieve only noun synsets for "dog"
dog_noun_synsets = wn.synsets("dog", pos=wn.NOUN)

for synset in dog_noun_synsets:
    print(synset)
```

## 3. Inspect a Synset

A synset provides several useful attributes:
- `name()` — the synset identifier.
- `definition()` — a short definition.
- `examples()` — example sentences, if available.
- `lemmas()` — lemma objects associated with the meaning.
- `pos()` — the part of speech.

```python
# Select the first synset for "car"
car_synsets = wn.synsets("car")

if car_synsets:
    car_synset = car_synsets[0]

    # Display basic information about the selected meaning
    print("Synset:", car_synset.name())
    print("Definition:", car_synset.definition())
    print("Examples:", car_synset.examples())
    print("Part of speech:", car_synset.pos())
    print("Lemmas:", [lemma.name() for lemma in car_synset.lemmas()])
else:
    print("No synsets found for 'car'.")
```

## 4. Explore Lemmas

A lemma is a word form associated with a particular synset. Lemmas can expose relationships such as antonyms.

Note that a lemma is not the same as a synset: a synset represents a meaning, while a lemma is one of the word forms linked to that meaning.

```python
# Inspect lemmas associated with the first synset of "good"
good_synsets = wn.synsets("good")

if good_synsets:
    good_synset = good_synsets[0]

    for lemma in good_synset.lemmas():
        # Print each lemma and its name
        print("Lemma:", lemma.name())
```

## 5. Hypernyms and Hyponyms

WordNet organizes many noun and verb meanings into semantic hierarchies.

- **Hypernym:** a more general concept (for example, `dog` is an animal).
- **Hyponym:** a more specific concept (for example, a particular kind of dog is a type of dog).

The exact results depend on which synset is selected.

```python
# Select a noun meaning of "dog"
dog_synsets = wn.synsets("dog", pos=wn.NOUN)

if dog_synsets:
    dog_synset = dog_synsets[0]

    # Get more general concepts
    print("Synset:", dog_synset.name())
    print("Hypernyms:")
    for hypernym in dog_synset.hypernyms():
        print(" -", hypernym.name(), ":", hypernym.definition())

    # Get more specific concepts
    print("\nHyponyms:")
    for hyponym in dog_synset.hyponyms()[:10]:
        print(" -", hyponym.name(), ":", hyponym.definition())
else:
    print("No noun synsets found for 'dog'.")
```

## 6. Find Antonyms

Some WordNet lemmas have antonym links. Antonyms are not available for every word or every meaning, so an empty result is normal.

```python
# Look for antonyms associated with the adjective "good"
good_adjective_synsets = wn.synsets("good", pos=wn.ADJ)

for synset in good_adjective_synsets:
    for lemma in synset.lemmas():
        # Display antonyms linked to this lemma, if any
        for antonym in lemma.antonyms():
            print(f"{lemma.name()} -> {antonym.name()}")
```

## 7. Explore Similarity Between Synsets

WordNet offers similarity measures for certain synset types. These measures compare semantic relationships in WordNet's hierarchy; they are not general-purpose contextual language similarity scores.

`path_similarity()` returns a score based on the shortest path between two synsets. Some comparisons may return `None`, especially when synsets are not connected in the relevant hierarchy.

For meaningful comparisons, choose synsets with compatible parts of speech.

```python
# Select noun synsets for two related words
cat_synsets = wn.synsets("cat", pos=wn.NOUN)
dog_synsets = wn.synsets("dog", pos=wn.NOUN)

if cat_synsets and dog_synsets:
    cat = cat_synsets[0]
    dog = dog_synsets[0]

    # Calculate path similarity between the selected synsets
    similarity = cat.path_similarity(dog)

    print("Cat synset:", cat.name())
    print("Dog synset:", dog.name())
    print("Path similarity:", similarity)
else:
    print("Could not find the required noun synsets.")
```

## 8. Find a Common Hypernym

A common hypernym is a more general concept shared by two synsets. It can help illustrate how WordNet relates concepts in a hierarchy.

```python
# Find the common hypernyms of the selected cat and dog synsets
if cat_synsets and dog_synsets:
    cat = cat_synsets[0]
    dog = dog_synsets[0]

    common_hypernyms = cat.lowest_common_hypernyms(dog)

    print("Lowest common hypernyms:")
    for synset in common_hypernyms:
        print(synset.name(), ":", synset.definition())
```
