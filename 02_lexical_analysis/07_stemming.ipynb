# 07 - Stemming in NLP

## Introduction

Stemming is a text normalization technique in Natural Language Processing (NLP).
It reduces words to their stems by removing or modifying word endings.

For example:
- `playing` → `play`
- `studies` → `studi` (depending on the algorithm)

A stem is not always a valid English word.

## 1. What Is Stemming?

Stemming is a text preprocessing technique that attempts to reduce
different forms of a word to a common stem.

For example, words such as:
- `playing`
- `played`
- `plays`

may be reduced to a similar stem.

Stemming is useful in information retrieval, search engines,
and some Natural Language Processing tasks.

### Stemming vs. Lemmatization

- **Stemming:** Uses rules to remove or modify word endings.
- **Lemmatization:** Attempts to find the dictionary base form
  of a word using linguistic information.

The output of a stemmer does not necessarily represent a valid word.

```python
# Import stemming algorithms from NLTK
from nltk.stem import (
    PorterStemmer,
    LancasterStemmer,
    RegexpStemmer,
    SnowballStemmer
)

# Define sample English words for the examples
words = [
    "information",
    "cooking",
    "activate",
    "relation",
    "playing",
    "studies",
    "connected"
]

# Display the original words
print("Original words:")
print(words)
```

## 2. PorterStemmer

PorterStemmer is a widely used stemming algorithm for English.

It applies a sequence of rules to reduce words to their stems.

In the following example, we apply PorterStemmer to a list of words
and compare the original words with their stems.

```python
# Create a PorterStemmer object
porter = PorterStemmer()

# Apply Porter stemming to each word
for word in words:
    stem = porter.stem(word)

    # Display the original word and its stem
    print(f"{word:12} -> {stem}")
```

## 3. LancasterStemmer

LancasterStemmer is another English stemming algorithm.

It can produce more aggressive reductions than PorterStemmer.
As a result, its output may be shorter and may differ from
the output of other stemming algorithms.

```python
# Create a LancasterStemmer object
lancaster = LancasterStemmer()

# Apply Lancaster stemming to each word
for word in words:
    stem = lancaster.stem(word)

    # Display the original word and its stem
    print(f"{word:12} -> {stem}")
```

## 4. Comparing PorterStemmer and LancasterStemmer

Different stemming algorithms may produce different results
for the same word.

In this section, we compare the outputs of PorterStemmer
and LancasterStemmer side by side.

```python
# Compare the output of two stemming algorithms
for word in words:
    porter_result = porter.stem(word)
    lancaster_result = lancaster.stem(word)

    # Print both results in a formatted row
    print(
        f"{word:12} | "
        f"Porter: {porter_result:10} | "
        f"Lancaster: {lancaster_result}"
    )
```

## 5. RegexpStemmer

RegexpStemmer allows us to define a regular expression pattern
that will be removed from a word.

In this example, we remove the suffix `ing`.

Important:
RegexpStemmer does not understand the meaning of a word.
It only applies the pattern that we specify.

Therefore, it may produce incorrect or unexpected stems.

```python
# Create a RegexpStemmer that removes the suffix "ing"
regexp_stemmer = RegexpStemmer("ing")

# Define sample words
regexp_words = [
    "cooking",
    "playing",
    "sing",
    "bring"
]

# Apply the regular-expression-based stemming
for word in regexp_words:
    stem = regexp_stemmer.stem(word)

    # Display the original word and the result
    print(f"{word:12} -> {stem}")
```

### 5.1. Using the `min` Parameter

The `min` parameter specifies the minimum word length
required for the regular expression to be applied.

In this example, the pattern is applied only when the word
contains at least four characters.

```python
# Create a RegexpStemmer with a minimum word length of 4
regexp_stemmer_min = RegexpStemmer("ing", min=4)

# Define words with different lengths
regexp_words = [
    "ing",
    "sing",
    "king",
    "cooking",
    "playing"
]

# Apply the pattern according to the minimum-length setting
for word in regexp_words:
    stem = regexp_stemmer_min.stem(word)

    # Display the original word and the result
    print(f"{word:12} -> {stem}")
```

## 6. SnowballStemmer

SnowballStemmer supports multiple languages.

Here, we use the English version to reduce words to their stems.

The Snowball algorithm is related to the Porter stemming family
and provides stemming implementations for several languages.

```python
# Create an English SnowballStemmer
snowball = SnowballStemmer("english")

# Apply Snowball stemming to each word
for word in words:
    stem = snowball.stem(word)

    # Display the original word and the result
    print(f"{word:12} -> {stem}")
```

### 6.1. Supported Languages

SnowballStemmer supports several languages.

We can inspect the languages available in the installed NLTK version
using the `languages` attribute.

```python
# Get the languages supported by SnowballStemmer
supported_languages = sorted(SnowballStemmer.languages)

# Display the number of supported languages
print("Number of supported languages:", len(supported_languages))

# Display the language names
print("Supported languages:")
print(supported_languages)
```

## 7. Comparing Three Stemming Algorithms

In this section, we compare:
- PorterStemmer
- LancasterStemmer
- SnowballStemmer

The same words are processed by all three algorithms.

This comparison helps us understand that different algorithms
may produce different stems for the same input.

```python
# Define additional words for a broader comparison
comparison_words = [
    "information",
    "cooking",
    "activate",
    "relation",
    "playing",
    "studies",
    "connected",
    "generously",
    "happiness"
]

# Compare all three stemming algorithms
for word in comparison_words:
    porter_result = porter.stem(word)
    lancaster_result = lancaster.stem(word)
    snowball_result = snowball.stem(word)

    # Print the original word and each algorithm's result
    print(f"Original:  {word}")
    print(f"Porter:    {porter_result}")
    print(f"Lancaster: {lancaster_result}")
    print(f"Snowball:  {snowball_result}")
    print("-" * 35)
```
