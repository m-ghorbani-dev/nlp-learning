## What Is a Text Corpus?
A **corpus** is a collection of texts or language data used for studying or processing language. A corpus may contain books, news articles, conversations, word lists, or texts in a particular language.

Examples include:
- A collection of English news articles
- A collection of personal names
- A collection of Persian texts
- A collection of texts grouped by topic

> Important distinction: a corpus is a collection of language data, while an NLP model is a system trained to perform a task such as classification or prediction.

## 2. Create a Custom Corpus

Loading text from a local file is useful when creating or experimenting with your own corpora.

First, we will create a small text file so that the example is self-contained and runnable. The file will be created in the notebook's current working directory.


```python
from pathlib import Path

corpus_dir = Path("my_corpus")
corpus_dir.mkdir(exist_ok=True)

sample_text = """Natural language processing works with human language.
A text corpus contains examples of language.
Researchers use corpora to study words and patterns."""

(corpus_dir / "corpus.txt").write_text(sample_text, encoding="utf-8")

print("Corpus file created:", corpus_dir / "corpus.txt")
print((corpus_dir / "corpus.txt").read_text(encoding="utf-8"))
```

### Reading a File with `Path`

In the example above:
- `Path("my_corpus")` specifies the directory path.
- `mkdir(exist_ok=True)` creates the directory and does not raise an error if it already exists.
- `write_text(..., encoding="utf-8")` saves the text using UTF-8 encoding.

For a single plain-text file, Python's built-in file-reading methods are sufficient. NLTK corpus readers provide additional features when working with multiple files or specific corpus formats.


```python
text = (corpus_dir / "corpus.txt").read_text(encoding="utf-8")
print(text.split())
```

## 3. Use `PlaintextCorpusReader`

`PlaintextCorpusReader` is suitable for reading one or more text files from a directory. In this example, we will read the file created in the previous section as a corpus.


```python
from pathlib import Path
import nltk
from nltk.corpus import PlaintextCorpusReader

# Create a corpus directory inside an existing NLTK data directory
corpus_dir = Path(nltk.data.path[0]) / "my_corpus"
corpus_dir.mkdir(parents=True, exist_ok=True)

# Create a sample text file
(corpus_dir / "corpus.txt").write_text(
    "Natural language processing works with human language.\n"
    "A text corpus contains examples of language.\n"
    "Researchers use corpora to study words and patterns.",
    encoding="utf-8"
)

# Initialize the corpus reader
reader = PlaintextCorpusReader(str(corpus_dir), r"corpus\.txt")

print("File IDs:", reader.fileids())
print("Words:", reader.words())
print("Sentences:", reader.sents())
```

Notes:
- `fileids()` returns the names of files in the corpus.
- `words()` returns the corpus words as a list-like collection.
- `sents()` returns sentences; the result depends on how sentence boundaries are identified for the corpus.

If your file path is different, update `corpus_dir` to match your directory.


## 4. Word Lists with `WordListCorpusReader`

`WordListCorpusReader` is useful for reading files in which each line contains one word or one list item. In this example, we will use `boy.txt` and `girl.txt`.

To keep the example runnable, we will create two small sample files.


```python
(corpus_dir / "boy.txt").write_text("Ali\nReza\nAmir\n", encoding="utf-8")
(corpus_dir / "girl.txt").write_text("Sara\nMina\nNeda\n", encoding="utf-8")

from nltk.corpus.reader import WordListCorpusReader

reader = WordListCorpusReader(str(corpus_dir), ["boy.txt", "girl.txt"])

print("All file IDs:", reader.fileids())
print("All names:", reader.words())
print("Names in boy.txt:", reader.words("boy.txt"))
print("Names in girl.txt:", reader.words("girl.txt"))
```

## 5. The Names Corpus: `names`

NLTK provides a names corpus split into two files: `male.txt` and `female.txt`.


```python
from nltk.corpus import names

print("File IDs:", names.fileids())
print("Number of names in male.txt:", len(names.words("male.txt")))
print("Number of names in female.txt:", len(names.words("female.txt")))

print("First 20 male names:", names.words("male.txt")[:20])
print("First 20 female names:", names.words("female.txt")[:20])
```

This corpus is useful for educational examples, such as examining letter frequencies in names or building a simple classifier. However, its categories have limitations and should not be treated as a complete representation of real-world identity or gender.


## 6. The English Words Corpus: `words`

The `words` corpus provides a list of English words. We can inspect its file IDs and view a few example entries.


```python
from nltk.corpus import words

print("File IDs:", words.fileids())
print("First 30 words:", words.words()[:30])
print("Total entries:", len(words.words()))
```

A word's presence in this corpus does not necessarily mean that it is common. The corpus may also contain word forms or entries that are not suitable for a particular application. In real-world projects, choose a word list that fits the task.


## 7. The Categorized Brown Corpus

Unlike a simple word list, the Brown corpus contains texts grouped by genre or text category. These categories are useful for sampling and comparing language across different styles.


```python
from nltk.corpus import brown

categories = brown.categories()
print("Categories:", categories)
print("Number of categories:", len(categories))

# Show a few words from one category.
category = categories[0]
print("Selected category:", category)
print("First 30 words:", brown.words(categories=category)[:30])
```

Use `brown.words(categories="news")` to retrieve words from the news category, or `brown.sents(categories="news")` to inspect its sentences. Check `brown.categories()` to see the available category names.


```python
news_words = brown.words(categories="news")
print("Number of words in news:", len(news_words))
print("First sentence in news:", brown.sents(categories="news")[0])
```
