# 22. Persian Stopwords

## Overview
Stopwords are common words that may contribute little to a particular analysis, such as some pronouns, particles, and prepositions. Removing them can simplify frequency analysis and some information-retrieval tasks, but they should not be removed automatically from every NLP pipeline.

This notebook demonstrates stopword handling for Persian text with Hazm and a small transparent custom list.

## 1. Import tokenizer and stopword resources

Hazm provides a Persian stopword list through `stopwords_list`. Resource availability can vary by installed version.

```python
# If needed, install in your environment:
# %pip install hazm

from hazm import word_tokenize, stopwords_list

persian_stopwords = set(stopwords_list())
print("Number of stopwords:", len(persian_stopwords))
print("Sample stopwords:", sorted(persian_stopwords)[:30])
```

## 2. Tokenize a Persian sentence

Start with a sentence that contains common function words.

```python
text = "من به یادگیری پردازش زبان طبیعی علاقه دارم و این موضوع برای من مهم است."
tokens = word_tokenize(text)

print("Tokens:", tokens)
```

## 3. Remove stopwords

The helper preserves token order and returns only tokens not found in the stopword set.

```python
def remove_stopwords(tokens, stopwords):
    return [token for token in tokens if token not in stopwords]

filtered_tokens = remove_stopwords(tokens, persian_stopwords)

print("Before:", tokens)
print("After :", filtered_tokens)
```

## 4. Compare custom and library stopwords

A custom list can be useful for a domain-specific task, but it should be documented and evaluated. The following list is intentionally small and only illustrates the mechanics.

```python
custom_stopwords = {"من", "به", "و", "این", "برای"}
custom_filtered = remove_stopwords(tokens, custom_stopwords)

print("Library list result:", filtered_tokens)
print("Custom list result :", custom_filtered)
```

## 5. Analyze a small collection of documents

This example compares token counts before and after stopword removal. It uses `Counter` to summarize token frequencies.

```python
from collections import Counter

documents = [
    "من به یادگیری ماشین علاقه دارم.",
    "پردازش زبان طبیعی برای تحلیل متن استفاده می‌شود.",
    "این روش برای دسته‌بندی متن مفید است.",
]

all_tokens = []
filtered_all_tokens = []

for document in documents:
    doc_tokens = word_tokenize(document)
    all_tokens.extend(doc_tokens)
    filtered_all_tokens.extend(remove_stopwords(doc_tokens, persian_stopwords))

print("Most common tokens before filtering:")
print(Counter(all_tokens).most_common(10))

print("\nMost common tokens after filtering:")
print(Counter(filtered_all_tokens).most_common(10))
```

## 6. When not to remove stopwords

Stopword removal can be harmful when function words carry task-relevant meaning. For example, negation words such as `نه` or `نیست` can reverse sentiment. Review the list and preserve important words for the target task.

## Summary

- Stopwords are task-dependent, not universally unimportant.
- Hazm provides a ready-to-use Persian stopword list.
- A custom list should be transparent and validated.
- Preserve negation and other meaningful tokens when needed.
