# 21. Persian Stemming and Lemmatization

## Overview
Stemming reduces a word to a shorter stem using heuristic rules. Lemmatization attempts to map a word to a canonical dictionary form. Persian morphology includes prefixes, suffixes, and compound constructions, so overly aggressive stemming can damage meaning.

This notebook introduces Hazm's stemmer and lemmatizer and compares their output on common Persian examples.

## 1. Import the tools

Hazm provides `Stemmer` and `Lemmatizer` classes for Persian text processing.

```python
# If needed, install in your environment:
# %pip install hazm

from hazm import Stemmer, Lemmatizer, word_tokenize

stemmer = Stemmer()
lemmatizer = Lemmatizer()
```

## 2. Apply stemming

Stemming typically removes some affixes. The resulting stem may not always be a standalone dictionary word.

```python
words = ["کتاب‌ها", "دانشجویان", "خانه‌ها", "رفتند", "زیباترین"]

for word in words:
    print(f"{word:<15} -> {stemmer.stem(word)}")
```

## 3. Apply lemmatization

Lemmatization uses lexical and morphological knowledge to return a canonical form where possible.

```python
words = ["می‌روم", "رفتند", "کتاب‌ها", "گفته‌ام", "بهترین"]

for word in words:
    print(f"{word:<15} -> {lemmatizer.lemmatize(word)}")
```

## 4. Compare stemmer and lemmatizer

The two tools use different strategies, so their outputs are not expected to match for every word.

```python
comparison_words = [
    "کتاب‌ها",
    "دانشجویان",
    "رفتند",
    "می‌نویسد",
    "خانه‌ها",
    "گفته‌ام",
]

for word in comparison_words:
    print(
        f"{word:<15} "
        f"stem={stemmer.stem(word):<15} "
        f"lemma={lemmatizer.lemmatize(word)}"
    )
```

## 5. Process a sentence

Tokenize the text first, then apply the selected morphological operation to each token. In a real pipeline, punctuation handling and token filtering should be designed for the target task.

```python
sentence = "دانشجویان کتاب‌ها را خواندند و درباره‌ی آن‌ها صحبت کردند."
tokens = word_tokenize(sentence)

stemmed_tokens = [stemmer.stem(token) for token in tokens]
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in tokens]

print("Tokens      :", tokens)
print("Stemmed     :", stemmed_tokens)
print("Lemmatized  :", lemmatized_tokens)
```

## 6. Choose the appropriate method

- Stemming can be useful when approximate word-form reduction is enough, such as some search applications.
- Lemmatization is preferable when a canonical lexical form is important.
- Always inspect outputs on domain-specific examples before using them in a model.

## Summary

- Stemming is generally heuristic and may produce non-words.
- Lemmatization aims for a canonical form and can use lexical information.
- Persian affixes and half-space forms make normalization and tokenization important.
- The best method depends on the downstream task and the quality of its outputs.
