# 20. Persian Tokenization

## Overview
Tokenization splits text into smaller units, such as sentences and words. Persian tokenization must account for punctuation, compound forms, and the zero-width non-joiner (half-space).

This notebook demonstrates sentence and word tokenization with Hazm and examines how tokenization behaves on Persian text.

## 1. Import Hazm tokenizers

Hazm includes tokenizers designed for Persian text.

```python
# If needed, install in your environment:
# %pip install hazm

from hazm import sent_tokenize, word_tokenize

text = (
    "پردازش زبان طبیعی بسیار جذاب است. "
    "آیا می‌خواهید تحلیل متن فارسی را یاد بگیرید؟ "
    "این مثال برای توکن‌سازی ساخته شده است!"
)

print(text)
```

## 2. Sentence tokenization

Sentence tokenization splits a document into sentence-level units.

```python
sentences = sent_tokenize(text)

for index, sentence in enumerate(sentences, start=1):
    print(f"{index}. {sentence}")
```

## 3. Word tokenization

Word tokenization splits each sentence into tokens. Depending on the tokenizer, punctuation may appear as separate tokens.

```python
tokens = word_tokenize(text)

print(tokens)
print("Token count:", len(tokens))
```

## 4. Inspect half-space behavior

The zero-width non-joiner (`\u200c`) appears in forms such as `می‌روم` and `کتاب‌ها`. Tokenization decisions can affect downstream frequency analysis and modeling.

```python
half_space_examples = [
    "می‌روم",
    "کتاب‌ها",
    "خانه‌ی ما",
    "نمی‌توانم",
]

for item in half_space_examples:
    print(f"{item!r} -> {word_tokenize(item)}")
```

## 5. Tokenize multiple documents

Represent each document as a list of tokens when preparing data for corpus-level analysis.

```python
documents = [
    "من به یادگیری ماشین علاقه دارم.",
    "پردازش زبان طبیعی یکی از شاخه‌های هوش مصنوعی است.",
    "تحلیل متن فارسی با ابزارهای مناسب ساده‌تر می‌شود.",
]

tokenized_documents = [word_tokenize(doc) for doc in documents]

for doc_id, doc_tokens in enumerate(tokenized_documents, start=1):
    print(f"Document {doc_id}: {doc_tokens}")
```

## 6. Compare raw and normalized tokenization

Normalization before tokenization can reduce variation caused by inconsistent spelling. Use the same preprocessing policy throughout a project.

```python
from hazm import Normalizer

normalizer = Normalizer()
raw = "اين   كتاب ها مفيد هستند!"
normalized = normalizer.normalize(raw)

print("Raw tokens       :", word_tokenize(raw))
print("Normalized text  :", normalized)
print("Normalized tokens:", word_tokenize(normalized))
```

## Summary

- Sentence tokenization creates sentence-level units.
- Word tokenization produces tokens for analysis.
- Half-space handling is important in Persian.
- Normalize consistently before tokenization when the task benefits from it.
