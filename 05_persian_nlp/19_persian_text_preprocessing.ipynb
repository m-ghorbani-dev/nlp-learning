# 19. Persian Text Preprocessing

## Overview
Persian text often contains Arabic/Persian character variants, inconsistent spacing, half-spaces, punctuation differences, and digits from different writing systems. This notebook introduces a practical preprocessing pipeline using Python and Hazm.

The goal is to normalize common writing variations while preserving the meaning and readability of Persian text.

## 1. Install and import dependencies

Install Hazm in your Python environment if it is not already available. In Jupyter, you can run the installation command in a terminal or use the commented notebook command below.

```python
# If needed, run this in a notebook cell:
# %pip install hazm

import re
from hazm import Normalizer

normalizer = Normalizer()
```

## 2. Inspect common Persian text issues

The sample includes Arabic variants of Persian letters, inconsistent whitespace, and punctuation spacing.

```python
raw_text = "مي خواهم   NLP را یاد بگيرم؛اين يك متنِ آزمايشي است!!!\nشماره ۱۲۳ و 123"

print("Original text:")
print(raw_text)
```

## 3. Normalize text with Hazm

Hazm's `Normalizer` handles common orthographic normalization. The exact transformations depend on the installed Hazm version and its options.

```python
normalized_text = normalizer.normalize(raw_text)

print("Normalized text:")
print(normalized_text)
```

## 4. Apply custom whitespace and punctuation cleanup

Normalization and cleanup are separate operations. This example standardizes whitespace and adds readable spacing around selected punctuation without removing Persian letters.

```python
def clean_persian_spacing(text):
    text = re.sub(r"[\t\r\f\v]+", " ", text)
    text = re.sub(r" *([،؛؟!]) *", r"\1 ", text)
    text = re.sub(r" +", " ", text)
    return text.strip()

cleaned_text = clean_persian_spacing(normalized_text)
print(cleaned_text)
```

## 5. Normalize Arabic and Persian digits

Some datasets mix Persian digits (`۰۱۲۳۴۵۶۷۸۹`), Arabic-Indic digits (`٠١٢٣٤٥٦٧٨٩`), and Western digits (`0123456789`). The helper below converts them to Persian digits.

```python
digit_translation = str.maketrans(
    "0123456789٠١٢٣٤٥٦٧٨٩",
    "۰۱۲۳۴۵۶۷۸۹۰۱۲۳۴۵۶۷۸۹",
)

def to_persian_digits(text):
    return text.translate(digit_translation)

digit_example = "Order 123، شماره ٤٥٦ و کد 789"
print(to_persian_digits(digit_example))
```

## 6. Build a reusable preprocessing function

A pipeline makes preprocessing consistent across documents. Keep the original text separately when traceability or auditing is important.

```python
def preprocess_persian_text(text):
    text = normalizer.normalize(text)
    text = clean_persian_spacing(text)
    return text

sample_documents = [
    "اين محصول خيلي خوب است!!!",
    "من   از کیفیت محصول راضی هستم؛ اما قیمت آن بالاست.",
    "پردازش زبان طبیعی (NLP) کاربردهای زیادی دارد.",
]

processed_documents = [preprocess_persian_text(doc) for doc in sample_documents]

for original, processed in zip(sample_documents, processed_documents):
    print("Original :", original)
    print("Processed:", processed)
    print("-" * 50)
```

## Summary

- Persian preprocessing should handle character variants and whitespace consistently.
- Hazm provides Persian-specific normalization.
- Custom cleanup should be conservative so that useful characters and meaning are preserved.
- Keep preprocessing rules consistent between training and inference.
